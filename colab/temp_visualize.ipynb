```python
import numpy as np
import matplotlib.pyplot as plt

# ===== CONFIG =====
file_path = "h_matrix_single.npy"  # Change to your file

# Load the .npy file
data = np.load(file_path)[:100,:100]  # Load only a subset for visualization

print(f"Loaded {file_path}")
print(f"Shape: {data.shape}")
print(f"Data type: {data.dtype}")

# Show basic statistics
print(f"Min: {np.nanmin(data)}")
print(f"Max: {np.nanmax(data)}")
print(f"Mean: {np.nanmean(data)}")

# Visualize as a heatmap
plt.imshow(data, cmap="viridis", interpolation="nearest")
plt.colorbar(label="Value")
plt.title(f"Visualization of {file_path}")
plt.show()
```
